- Read the dataset labelled features.csv.
- Create a ``` lambda ``` function that checks if displacement $>$ 200
- Create a new list that shows true if displacement $>$ 200 and false <br>
if displacement $\leq$ 200.
- Count how many cars have displacement $> 200$.

To get full credit use either the apply or the map function <br>
(see Lectures 7 and 8)

````python
# Write your answer here:

#reading the dataset
carfeatures = pd.read_csv("data/features.csv")

#create lambda 
checkdisplacement = lambda x: x > 200

#new list for true/ false
carfeatures['above200'] = carfeatures['displacement'].apply(checkdisplacement)
print(carfeatures['above200'])

#counting cars with displacement>200
count_cars = carfeatures['above200'].sum()
print(count_cars)
````

(b) Subsetting Data

<font size = "3" >

- Read the dataset labelled features.csv.
- Define a global variable called ```threshold``` that takes the <br>
value 2588
- Subset the data to get one dataset with weight $>$ threshold and <br>
another with weight $\le$ 2588. Use the ```.query()``` function.
- Compute the mean acceleration for each of these datasets.<br>

Note: Review how to define a gloabl variable inside ```.query()```
 

````python
# Write your answer here:

carfeatures = pd.read_csv("data/features.csv")
threshold = 2588

#subsetting with .query()
above_threshold = carfeatures.query("weight > @threshold")
below_threshold = carfeatures.query("weight <= 2588") 

#computing mean accelerations
above_mean_acc = above_threshold['acceleration'].mean()
below_mean_acc = below_threshold['acceleration'].mean()

#printing means
print(above_mean_acc)
print(below_mean_acc)
````

(d) Use queries + global variables

<font size = "3">

- Read the "fifa23_players_basic.csv" dataset
- Compute the mean of the column "Wage(in Euro)" <br>
and store it as a variable
- Subset the players whose value is above the mean value <br>
  using ".query()"


HINT: Remember that there is a special syntax for <br>
when the column names have spaces (see Lecture 10) <br>
Reference the column as ``` `Wage(in Euro)`    ``` inside .query()

````python
# Write your answer here:
fifa = pd.read_csv("data/fifa23_players_basic.csv") 

#computing the mean of column Wage
wage_mean = fifa['Wage(in Euro)'].mean()

#subsetting above average wage players
above_mean_wage = fifa.query("`Wage(in Euro)` > @wage_mean")
print(above_mean_wage)
````

(e) Plot by Category

<font size = "3">

Read the "data/wdi_urban.csv" dataset.

  This contains data from 1980 and 2020.

- **Create a list with the unique values of "year"**

- **Create a plot with a ...**
    - Histogram of "prop_urbanpopulation" for each year
    - Set the histogram option `alpha = 0.5` for transparency
    - All the histograms should be overlapping
    - Create a legend with the list of years
    - Label the axes and the title


````python
# Write your answer here:

urban = pd.read_csv("data/wdi_urban.csv") 

list_unique_year = pd.unique(urban["year"])

plt.figure(figsize=(12,6))  

for year in list_unique_year:
    subset = urban[urban["year"] == year]["prop_urbanpopulation"]
    plt.hist(subset, alpha = 0.5)

plt.xlabel("Urban Population")
plt.ylabel("Frequency")
plt.legend(labels = list_unique_year,
           title = "Year")
plt.title("Urban Population Frequency per Year")
plt.show()
````

```python
# Importing Libraries

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import ticker
import matplotlib.dates as mdates
```

```python
# Import Dataset

constructor_results = pd.read_csv("data_raw/constructor_results.csv") 
constructors = pd.read_csv("data_raw/constructors.csv") 
results = pd.read_csv("data_raw/results.csv")
drivers = pd.read_csv("data_raw/drivers.csv") 
```

```python
#merging datasets

driver_results = results.merge(
    drivers[['driverId', 'driverRef', 'forename', 'surname', 'nationality']],
    on='driverId',
    how='left'
)

merged_constructor_results = constructor_results.merge(
    constructors[['constructorId', 'name', 'nationality']],
    on='constructorId',
    how='left'
)
```
